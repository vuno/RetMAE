```python
import matplotlib.pyplot as plt
import pandas as pd

# Set global font and tick size
plt.rcParams["font.family"] = "Times New Roman"
plt.rcParams["xtick.direction"] = "in"
plt.rcParams["ytick.direction"] = "in"
plt.rcParams["xtick.labelsize"] = 12
plt.rcParams["ytick.labelsize"] = 12

# 데이터 로드
fundus_df = pd.read_csv("high_freq_fundus.csv")
imgnet_df = pd.read_csv("high_freq_imgnet.csv")

# 서브플롯 생성
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(12, 3.5), sharey=True)

# --- Q25 ---
ax1.bar(
    fundus_df["ratio"],
    fundus_df["density_q25"],
    width=0.01,
    alpha=0.7,
    label="Fundus",
    color="#0D5EA6",
)
ax1.bar(
    imgnet_df["ratio"],
    imgnet_df["density_q25"],
    width=0.01,
    alpha=0.5,
    label="ImageNet",
    color="#E14434",
)
ax1.set_title("1st quartile", fontsize=14)
ax1.set_ylabel("Probability Density", fontsize=14, labelpad=10)
ax1.tick_params(axis="both", direction="in")

# --- Q50 (Median) ---
ax2.bar(
    fundus_df["ratio"],
    fundus_df["density_median"],
    width=0.01,
    alpha=0.7,
    label="Fundus",
    color="#0D5EA6",
)
ax2.bar(
    imgnet_df["ratio"],
    imgnet_df["density_median"],
    width=0.01,
    alpha=0.5,
    label="ImageNet",
    color="#E14434",
)
ax2.set_title("Median", fontsize=14)
ax2.set_xlabel("High-frequency ratio", fontsize=14, labelpad=10)
ax2.tick_params(axis="both", direction="in")

# --- 범례 중앙 위 ---
ax2.legend(fontsize=12, loc="lower center", bbox_to_anchor=(0.5, 1.18), ncol=2)

# --- Q75 ---
ax3.bar(
    fundus_df["ratio"],
    fundus_df["density_q75"],
    width=0.01,
    alpha=0.7,
    label="Fundus",
    color="#0D5EA6",
)
ax3.bar(
    imgnet_df["ratio"],
    imgnet_df["density_q75"],
    width=0.01,
    alpha=0.5,
    label="ImageNet",
    color="#E14434",
)
ax3.set_title("3rd quartile", fontsize=14)
ax3.tick_params(axis="both", direction="in")

# 저장
plt.tight_layout()
plt.savefig("plot_high_freq.pdf", format="pdf")
plt.close(fig)
```
