```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1) 폰트 설정: Times New Roman
plt.rc("font", family="serif", serif=["Times New Roman"], size=12)
plt.rc("mathtext", fontset="cm")

# 2) 엑셀 파일 경로
file_path = "abl_visible_ratio.xlsx"

# 3) Top high-freq. 시트에서 데이터 읽기
df = pd.read_excel(file_path, sheet_name="Max high-freq.")
df = df[["Ratio", "AR"]].sort_values("Ratio")

# 4) MAE (our impl.) 시트에서 데이터 읽기
mae_df = pd.read_excel(file_path, sheet_name="MAE (our impl.)")
mae_df = mae_df[["Ratio", "AR"]].sort_values("Ratio")

# 5) 세로로 긴(portrait) 그림판 생성
fig, ax = plt.subplots(figsize=(5, 3.5))

# 6) Top high-freq. 플롯
colors = plt.cm.Set1(np.linspace(0, 1, len(df)))
ax.plot(
    df["Ratio"],
    df["AR"],
    linestyle="-",
    marker="o",
    markersize=5,
    color="#0D5EA6",
    label="Max high-freq.",
)

# 7) MAE (our impl.) 플롯 (회색)
ax.plot(
    mae_df["Ratio"],
    mae_df["AR"],
    linestyle="-",
    marker="o",
    markersize=5,
    color="#E14434",
    label="MAE (our impl.)",
)

# 빨간색 그래프의 최고 포인트에 별표 추가
best_df_idx = df["AR"].idxmax()
best_df_ratio = df.loc[best_df_idx, "Ratio"]
best_df_value = df.loc[best_df_idx, "AR"]
ax.plot(
    best_df_ratio,
    best_df_value - 0.02,
    marker="*",
    markersize=14,
    color="#0D5EA6",
    label=None,
    zorder=5,
)

# 9) xticks를 Top high-freq. Ratio 값으로만 설정
ax.set_xticks(df["Ratio"])

# 10) 축 레이블 설정
ax.set_xlabel("Visible Ratio", fontsize=16, labelpad=10)
ax.set_ylabel("AUROC (%)", fontsize=16, labelpad=10)

# 11) x-grid는 xticks 위치에서만, y-grid는 전체 y축 major tick 위치에서
ax.grid(axis="x", linestyle="--")  # x-grid: Ratio 위치에서만
ax.grid(axis="y", linestyle="--")  # y-grid: y축 major tick 위치에서

# 12) tick label 크기 조절
ax.tick_params(axis="both", which="major", labelsize=14, pad=8, direction="in")

# 13) 범례 추가
ax.legend(fontsize=11, loc="upper left")

# 14) 축 범위 설정
ax.set_xlim(0, 0.27)
ax.set_ylim(80, 87)

# 15) 레이아웃 정리 및 PDF 저장
plt.tight_layout()
plt.savefig("abl_visible_ratio.pdf", format="pdf")
```
