```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.family"] = "Times New Roman"

# Data
strategies = ["Full", "Random (25% masking)", "Low-freq.", "High-freq."]
cka_values = [1.000, 0.997, 0.986, 0.537]
dr_values = [82.3, 82.8, 79.7, 83.7]
colors = ["#D2C1B6", "#D2C1B6", "#D2C1B6", "#0D5EA6"]

# Subplots with shared x-axis
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(6, 6), sharex=True)

# Top plot: CKA
ax1.bar(strategies, cka_values, color=colors, zorder=2, width=0.5)
ax1.set_ylabel("CKA Similarity", fontsize=14, labelpad=10)  # Increase labelpad
ax1.set_ylim(0.4, 1.05)
ax1.grid(axis="y", linestyle="--", zorder=0)
ax1.tick_params(axis="y", labelsize=12, pad=8, direction="in")  # Increase tick-label pad
ax1.tick_params(axis="x", bottom=False, labelbottom=False)

# Bottom plot: DR
ax2.bar(strategies, dr_values, color=colors, zorder=2, width=0.5)
ax2.set_xlabel("Inference Strategy", fontsize=14, labelpad=10)  # Increase labelpad
ax2.set_ylabel("AUROC (%)", fontsize=14, labelpad=10)  # Increase labelpad
ax2.set_ylim(78, 85)
ax2.grid(axis="y", linestyle="--", zorder=0)
ax2.tick_params(axis="x", labelsize=12, pad=8, direction="in")  # Increase tick-label pad
ax2.tick_params(axis="y", labelsize=12, pad=8, direction="in")  # Increase tick-label pad

fig.tight_layout(h_pad=2.0)
fig.savefig("obs_cka_perfom.pdf", format="pdf")
plt.show()
```
