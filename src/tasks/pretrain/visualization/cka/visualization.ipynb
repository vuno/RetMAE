```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# 폰트 설정 - Times New Roman
plt.rcParams["font.family"] = "Times New Roman"
plt.rcParams["font.size"] = 12
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 12
plt.rcParams["xtick.labelsize"] = 10
plt.rcParams["ytick.labelsize"] = 10
plt.rcParams["legend.fontsize"] = 10
```

```python
import matplotlib.pyplot as plt

# 제목과 파일 경로
title = "CKA similarity between full tokens and masked tokens"
filepath = (
    "your_path/src/tasks/pretrain/visualization/cka/cka_total.xlsx"
)

# 선택할 마스킹 방식과 라벨
selection = [
    ("random_masking", 0.0),
    ("random_masking", 0.3),
    ("random_masking", 0.7),
    ("high_freq_masking", 1.0),
    ("low_freq_masking", 1.0),
]
labels = {
    "random_masking (ratio 0)": "Full tokens",
    "random_masking (ratio 0.3)": "30% random masked",
    "random_masking (ratio 0.7)": "70% random masked",
    "high_freq_masking (ratio 1)": "All high-freq. masked (30% masking)",
    "low_freq_masking (ratio 1)": "All low-freq. masked (70% masking)",
}

# 그래프 생성
fig, ax = plt.subplots(figsize=(6, 5))

# 데이터 로드 및 처리
xlsx = pd.read_excel(filepath, sheet_name=None)
combined = []
for sheet_name, df in xlsx.items():
    df = df.rename(columns={df.columns[0]: "masking_ratio"})
    melted = df.melt(id_vars=["masking_ratio"], var_name="block", value_name="cka")
    melted["mask_method"] = sheet_name
    combined.append(melted)
combined_df = pd.concat(combined, ignore_index=True)

# block 번호 추출
all_blocks = combined_df["block"].str.extract(r"(\d+)").astype(int).iloc[:, 0].unique()
all_blocks_sorted = sorted(all_blocks)

# plot
legend_handles = {}
for method, ratio in selection:
    df_sel = combined_df[
        (combined_df["mask_method"] == method) & (combined_df["masking_ratio"] == ratio)
    ]
    df_sel = df_sel.copy()
    df_sel["block_num"] = df_sel["block"].str.extract(r"(\d+)").astype(int)
    df_sel = df_sel.sort_values("block_num")
    key = f"{method} (ratio {int(ratio) if ratio.is_integer() else ratio})"

    (line,) = ax.plot(
        df_sel["block_num"],
        df_sel["cka"],
        label=labels[key],
        marker="o" if ratio != 0.0 else None,
        linestyle="-" if ratio != 0.0 else "--",
        linewidth=1.5,
        markersize=5,
    )
    legend_handles[labels[key]] = line

# 스타일 설정
ax.set_title(title, pad=5)
ax.set_xlabel("Block")
ax.set_ylabel("CKA Similarity")
ax.set_xticks(all_blocks_sorted)
ax.grid(True, which="both", linestyle=":", linewidth=0.5)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.tick_params(labelleft=True)

# 그래프 안쪽에 자동 배치되는 Legend
legend = ax.legend(
    handles=list(legend_handles.values()),
    labels=list(legend_handles.keys()),
    loc="best",  # 자동으로 가장 적절한 위치 선택
    frameon=True,
    fontsize="small",
)

# Legend 스타일
legend.get_frame().set_edgecolor("lightgray")
legend.get_frame().set_linewidth(1.0)
legend.get_frame().set_boxstyle("round")
legend.get_frame().set_facecolor("#f9f9f9")

plt.tight_layout(rect=[0, 0.1, 1, 1])
plt.show()
```
