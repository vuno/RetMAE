```python
import matplotlib.pyplot as plt
import pandas as pd

# 데이터 로드
df = pd.read_csv(
    "your_path/src/tasks/pretrain/visualization/cka/cka_analysis_along_epoch.csv"
)

# Set font to Times New Roman
plt.rcParams["font.family"] = "Times New Roman"


# 데이터 변형
melted = df.melt(id_vars=["Infer"], var_name="epoch", value_name="cka")
melted["epoch_num"] = melted["epoch"].str.extract(r"epoch (\d+)").astype(int)

# 스타일 매핑
labels_map = {"high-freq.": "High-frequency Masking", "low-freq.": "Low-frequency Masking"}
colors = {"high-freq.": "#FF6666", "low-freq.": "#6699FF"}
linestyles = {"high-freq.": "--", "low-freq.": "-"}

# 플롯 생성
fig, ax = plt.subplots(figsize=(6, 5))

# 기존 라인 플롯
for method, group in melted.groupby("Infer"):
    ax.plot(
        group["epoch_num"],
        group["cka"],
        label=labels_map.get(method, method),
        marker="o",
        linestyle=linestyles.get(method, "-"),
        linewidth=1.5,
        markersize=5,
        color=colors.get(method),
    )

# Full Token 수평선 추가 (작은 점선)
ax.axhline(y=1.0, color="gray", linestyle=":", linewidth=2, label="Full Token")

# 레이블 및 제목
ax.set_xlabel("Epoch", fontsize=18, labelpad=12)
ax.set_ylabel("CKA Similarity", fontsize=18, labelpad=12)
ax.tick_params(axis="both", which="major", pad=8, labelsize=14)

# 축, 그리드 설정
ax.set_xticks(sorted(melted["epoch_num"].unique()))
ax.grid(True, which="both", linestyle=":", linewidth=0.5)

# 모든 테두리(스파인) 표시
for spine in ax.spines.values():
    spine.set_visible(True)
    spine.set_linewidth(1.0)

# 범례
legend = ax.legend(loc="best", frameon=True, fontsize=14)
legend.get_frame().set_edgecolor("lightgray")
legend.get_frame().set_linewidth(1.0)
legend.get_frame().set_boxstyle("round")
legend.get_frame().set_facecolor("#f9f9f9")

plt.tight_layout()
plt.show()
```
